## Reproduce one account directly from the engine

Monthly dollar orders use the previous close. Old holdings earn execution-day returns before the contribution arrives at the close. No action-file dividends are added to total-return units.


```python
from roaring_returns.data import load_prices
from roaring_returns.engine import BacktestConfig, backtest, summarize
from roaring_returns.experiments import window_with_anchor
prices, manifest = load_prices(root)
dev = window_with_anchor(prices, '2005-01-01', '2014-12-31')
result = backtest(dev, BacktestConfig('min_sales_band'))
display(pd.Series(summarize(result)))
display(result.events.head())
```

## Limits and next experiments

This is not a pay-date cash-dividend or whole-share simulator. The single public provider has not been independently certified. Fees reduce buys to avoid borrowing, so realized fills need not regain the frictionless signal-state band. Validation (2015–2019) and final retrospective evaluation (2020–2025) are separate explicit runs; keep final evaluation out of parameter selection.
